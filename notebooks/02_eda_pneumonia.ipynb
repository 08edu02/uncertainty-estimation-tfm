# EDA — escenario de imágenes (PneumoniaMNIST)

Exploración de PneumoniaMNIST (MedMNIST v2). Igual que en el notebook tabular, aquí no hay lógica crítica: se usan las funciones de `uq.data.pneumonia` y `uq.eda`.

Antes de ejecutarlo, descarga los datos:

```bash
uv run python scripts/download_data.py
```

````python
from IPython.display import Image, display

from uq.data import pneumonia
from uq.eda import figures, summary
from uq.settings import figures_dir

imagenes, etiquetas = pneumonia.load_split("train")
print(imagenes.shape, imagenes.dtype, etiquetas.shape)
````

````python
# Resumen por conjunto (el mismo que se guarda en results/eda/pneumonia.json).
resumen = {
    split: summary.image_summary(*pneumonia.load_split(split), split=split)
    for split in pneumonia.SPLITS
}

for split, datos in resumen.items():
    print(
        f"{split:>5}: {datos['n_samples']:>5} imágenes | "
        f"clase 1 = {datos['class_rates']['1']:.4f} | forma {datos['shape']}"
    )
````

````python
# Ojo: la proporción de clase positiva NO es igual en train/val (0.742) que en
# test (0.625). Hay que tenerlo en cuenta al interpretar exactitud y umbrales.
import pandas as pd

pd.DataFrame({split: datos["class_counts"] for split, datos in resumen.items()}).T.rename_axis(
    "split"
)
````

````python
# Distribución de intensidades: ¿hay imágenes muy oscuras o saturadas?
print("min", imagenes.min(), "max", imagenes.max(), "media", imagenes.mean().round(2))
````

````python
figures.plot_image_samples(
    imagenes,
    etiquetas,
    figures_dir() / "pneumonia_samples.png",
    class_names=pneumonia.CLASS_NAMES,
)
display(Image(filename=str(figures_dir() / "pneumonia_samples.png")))
````
